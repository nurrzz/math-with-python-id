# Median

Median adalah nilai tengah dari data yang sudah diurutkan.

Median berguna saat kita ingin melihat pusat data tanpa terlalu dipengaruhi oleh nilai yang sangat besar atau sangat kecil.

## Ide Matematika

Langkah mencari median:

1. Urutkan data dari kecil ke besar.
2. Jika jumlah data ganjil, ambil nilai yang tepat di tengah.
3. Jika jumlah data genap, ambil rata-rata dari dua nilai tengah.

Contoh data ganjil:

```text
70, 80, 90
median = 80
```

Contoh data genap:

```text
70, 80, 90, 100
dua nilai tengah = 80 dan 90
median = (80 + 90) / 2 = 85
```

````python
def median_tanpa_library(data):
    if len(data) == 0:
        raise ValueError("Data tidak boleh kosong.")

    data_urut = sorted(data)
    n = len(data_urut)
    tengah = n // 2

    if n % 2 == 1:
        return data_urut[tengah]

    return (data_urut[tengah - 1] + data_urut[tengah]) / 2


median_tanpa_library([70, 80, 90, 100])
````

## Dengan Library

NumPy menyediakan fungsi `median()` untuk menghitung median.

````python
import numpy as np

data = [70, 80, 90, 100]
np.median(data)
````

## Perbandingan dengan Rata-rata

Median sering lebih tahan terhadap nilai ekstrem.

```text
data = 10, 20, 30, 40, 1000
```

Rata-rata akan naik jauh karena ada `1000`, sedangkan median tetap menunjukkan nilai tengah data.

````python
data = [10, 20, 30, 40, 1000]

print("Rata-rata:", np.mean(data))
print("Median:", np.median(data))
````
