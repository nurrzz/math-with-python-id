# Rata-rata / Mean

Rata-rata adalah nilai yang mewakili sekumpulan data dengan cara membagi jumlah semua data dengan banyaknya data.

Konsep ini sering dipakai saat kita ingin melihat gambaran umum dari data, misalnya nilai ujian, pengeluaran bulanan, atau jumlah pengunjung harian.

## Ide Matematika

Rumus rata-rata:

```text
rata-rata = jumlah semua data / banyak data
```

Contoh:

```text
data = 70, 80, 90
jumlah semua data = 70 + 80 + 90 = 240
banyak data = 3
rata-rata = 240 / 3 = 80
```

## Dari Rumus ke Algoritma

Kalau diterjemahkan ke langkah berpikir komputer:

1. Simpan data dalam list.
2. Jumlahkan semua nilai.
3. Hitung banyak data.
4. Bagi total dengan banyak data.

````python
data = [70, 80, 90]

total = 0

for nilai in data:
    total += nilai

rata_rata = total / len(data)
rata_rata
````

## Versi Fungsi

Agar bisa dipakai berkali-kali, prosesnya bisa dibungkus menjadi fungsi.

````python
def mean_tanpa_library(data):
    if len(data) == 0:
        raise ValueError("Data tidak boleh kosong.")

    total = 0

    for nilai in data:
        total += nilai

    return total / len(data)


mean_tanpa_library([70, 80, 90])
````

## Dengan Library

NumPy menyediakan fungsi `mean()` untuk menghitung rata-rata secara langsung.

````python
import numpy as np

data = [70, 80, 90]
np.mean(data)
````

## Perbandingan

| Pendekatan | Yang Terlihat | Cocok Untuk |
| --- | --- | --- |
| Tanpa library | Proses menjumlahkan dan membagi data | Memahami konsep |
| Dengan library | Cara ringkas yang umum dipakai | Analisis data dan project nyata |

Versi manual membantu kita memahami isi rumusnya. Versi library membantu saat data sudah lebih besar atau pekerjaan perlu lebih praktis.

## Catatan

Rata-rata mudah dihitung, tapi tidak selalu cukup untuk memahami data.

```text
data A = 70, 80, 90
data B = 10, 80, 150
```

Keduanya punya rata-rata `80`, tetapi penyebaran datanya berbeda. Karena itu, kita juga perlu memahami median, modus, varians, dan standar deviasi.
