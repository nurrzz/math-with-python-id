# Varians dan Standar Deviasi

Varians dan standar deviasi digunakan untuk melihat seberapa menyebar data dari rata-ratanya.

Kalau rata-rata memberi gambaran pusat data, standar deviasi membantu menjawab: nilai-nilainya dekat dengan pusat, atau tersebar jauh?

## Ide Matematika

Langkah umum menghitung varians populasi:

1. Hitung rata-rata.
2. Hitung selisih setiap nilai dari rata-rata.
3. Kuadratkan setiap selisih.
4. Ambil rata-rata dari hasil kuadrat tersebut.

Standar deviasi adalah akar kuadrat dari varians.

```text
standar deviasi = akar(varians)
```

````python
def mean_tanpa_library(data):
    return sum(data) / len(data)


def varians_populasi_tanpa_library(data):
    if len(data) == 0:
        raise ValueError("Data tidak boleh kosong.")

    rata_rata = mean_tanpa_library(data)
    total_selisih_kuadrat = 0

    for nilai in data:
        selisih = nilai - rata_rata
        total_selisih_kuadrat += selisih ** 2

    return total_selisih_kuadrat / len(data)


data = [70, 80, 90]
varians_populasi_tanpa_library(data)
````

````python
def standar_deviasi_populasi_tanpa_library(data):
    varians = varians_populasi_tanpa_library(data)
    return varians ** 0.5


standar_deviasi_populasi_tanpa_library([70, 80, 90])
````

## Dengan Library

NumPy menyediakan `var()` untuk varians dan `std()` untuk standar deviasi.

````python
import numpy as np

data = [70, 80, 90]

print("Varians:", np.var(data))
print("Standar deviasi:", np.std(data))
````

## Cara Membaca Hasil

Semakin kecil standar deviasi, semakin dekat data dengan rata-ratanya.

Semakin besar standar deviasi, semakin menyebar data dari rata-ratanya.

````python
data_rapat = [78, 79, 80, 81, 82]
data_menyebar = [40, 60, 80, 100, 120]

print("Rata-rata data rapat:", np.mean(data_rapat))
print("Standar deviasi data rapat:", np.std(data_rapat))

print("Rata-rata data menyebar:", np.mean(data_menyebar))
print("Standar deviasi data menyebar:", np.std(data_menyebar))
````
