# Modus

Modus adalah nilai yang paling sering muncul dalam data.

Modus berguna saat kita ingin tahu nilai yang paling umum, misalnya ukuran sepatu yang paling banyak dibeli atau kategori jawaban yang paling sering dipilih.

## Ide Matematika

Untuk mencari modus, kita menghitung frekuensi kemunculan setiap nilai.

Contoh:

```text
data = 70, 80, 80, 90
70 muncul 1 kali
80 muncul 2 kali
90 muncul 1 kali
modus = 80
```

````python
def modus_tanpa_library(data):
    if len(data) == 0:
        raise ValueError("Data tidak boleh kosong.")

    frekuensi = {}

    for nilai in data:
        if nilai not in frekuensi:
            frekuensi[nilai] = 0
        frekuensi[nilai] += 1

    kemunculan_terbanyak = max(frekuensi.values())
    modus = []

    for nilai, jumlah in frekuensi.items():
        if jumlah == kemunculan_terbanyak:
            modus.append(nilai)

    return modus


modus_tanpa_library([70, 80, 80, 90])
````

## Dengan Library

Python punya module bawaan `statistics`. Untuk modus tunggal, kita bisa memakai `statistics.mode()`. Jika ingin semua nilai yang sama-sama paling sering muncul, gunakan `statistics.multimode()`.

````python
import statistics

data = [70, 80, 80, 90]

print(statistics.mode(data))
print(statistics.multimode(data))
````

## Data Bisa Punya Lebih dari Satu Modus

Kadang ada beberapa nilai yang sama-sama paling sering muncul. Data seperti ini disebut multimodal.

````python
data = [70, 70, 80, 80, 90]

print(modus_tanpa_library(data))
print(statistics.multimode(data))
````

## Catatan

Modus tidak selalu cocok untuk semua jenis data numerik, tapi sangat berguna untuk data kategori.

Contoh data kategori:

```text
warna favorit = merah, biru, biru, hijau
modus = biru
```
