# Persamaan Kuadrat

Persamaan kuadrat adalah persamaan yang memiliki variabel berpangkat dua. Bentuk umumnya:

```text
ax² + bx + c = 0
```

Contoh:

```text
x² - 5x + 6 = 0
```

## Ide Faktorisasi

Persamaan:

```text
x² - 5x + 6 = 0
```

Bisa difaktorkan menjadi:

```text
(x - 2)(x - 3) = 0
```

Artinya, nilai `x` yang membuat persamaan benar adalah:

```text
x = 2 atau x = 3
```

````python
def nilai_kuadrat(x):
    return x**2 - 5*x + 6


print(nilai_kuadrat(2))
print(nilai_kuadrat(3))
````

## Rumus Kuadrat

Untuk bentuk umum:

```text
ax² + bx + c = 0
```

Akar-akarnya bisa dihitung dengan rumus:

```text
x = (-b ± √(b² - 4ac)) / 2a
```

Bagian `b² - 4ac` disebut diskriminan.

````python
def akar_kuadrat(a, b, c):
    if a == 0:
        raise ValueError("a tidak boleh 0 untuk persamaan kuadrat.")

    diskriminan = b**2 - 4*a*c

    if diskriminan < 0:
        return None

    akar_1 = (-b + diskriminan**0.5) / (2*a)
    akar_2 = (-b - diskriminan**0.5) / (2*a)

    return akar_1, akar_2


akar_kuadrat(1, -5, 6)
````

## Dengan Library: SymPy

SymPy bisa menyelesaikan persamaan kuadrat secara simbolik.

````python
import sympy as sp

x = sp.Symbol('x')
persamaan = sp.Eq(x**2 - 5*x + 6, 0)

sp.solve(persamaan, x)
````

## Catatan tentang Diskriminan

Diskriminan membantu membaca jenis akar:

| Diskriminan | Makna |
| --- | --- |
| Lebih dari 0 | Ada dua akar real |
| Sama dengan 0 | Ada satu akar real kembar |
| Kurang dari 0 | Tidak ada akar real |

Untuk awal belajar, cukup pahami bahwa diskriminan memberi petunjuk tentang bentuk solusi.
