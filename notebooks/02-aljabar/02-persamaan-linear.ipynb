# Persamaan Linear

Persamaan linear adalah persamaan dengan variabel berpangkat satu. Bentuk sederhananya seperti ini:

```text
ax + b = c
```

Tujuannya biasanya mencari nilai `x` yang membuat persamaan menjadi benar.

## Contoh Manual

Misalnya:

```text
2x + 3 = 11
```

Langkahnya:

```text
2x + 3 = 11
2x = 11 - 3
2x = 8
x = 8 / 2
x = 4
```

````python
x = 4
2 * x + 3 == 11
````

## Dari Bentuk Umum ke Kode

Untuk persamaan:

```text
ax + b = c
```

Kita bisa memindahkan `b` ke kanan, lalu membagi dengan `a`:

```text
x = (c - b) / a
```

````python
def selesaikan_linear(a, b, c):
    if a == 0:
        raise ValueError("a tidak boleh 0 untuk persamaan linear satu variabel.")

    return (c - b) / a


selesaikan_linear(2, 3, 11)
````

## Dengan Library: SymPy

SymPy bisa menyelesaikan persamaan dengan bentuk yang lebih dekat ke penulisan matematika.

````python
import sympy as sp

x = sp.Symbol('x')
persamaan = sp.Eq(2 * x + 3, 11)

sp.solve(persamaan, x)
````

## Membaca Hasil

Hasil dari `sp.solve()` biasanya berupa list. Untuk contoh ini, hasilnya `[4]`, artinya nilai `x` yang memenuhi persamaan adalah `4`.

## Catatan

Versi manual bagus untuk memahami proses memindahkan ruas dan membagi koefisien. Versi SymPy bagus saat persamaan mulai lebih panjang atau saat kita ingin fokus pada bentuk matematikanya.
