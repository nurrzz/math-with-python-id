# Sistem Persamaan Linear

Sistem persamaan linear berisi lebih dari satu persamaan yang harus benar pada saat yang sama.

Contoh:

```text
x + y = 10
x - y = 2
```

Tujuannya mencari nilai `x` dan `y` yang memenuhi dua persamaan tersebut.

## Ide Eliminasi

Dari contoh:

```text
x + y = 10
x - y = 2
```

Jika kedua persamaan dijumlahkan:

```text
(x + y) + (x - y) = 10 + 2
2x = 12
x = 6
```

Setelah tahu `x = 6`, masukkan ke persamaan pertama:

```text
6 + y = 10
y = 4
```

````python
x = 6
y = 4

print(x + y == 10)
print(x - y == 2)
````

## Bentuk Matriks

Sistem persamaan linear sering ditulis dalam bentuk matriks.

```text
x + y = 10
x - y = 2
```

Koefisiennya bisa ditulis sebagai:

```text
[[1,  1],
 [1, -1]]
```

Hasil di ruas kanan:

```text
[10, 2]
```

````python
import numpy as np

A = np.array([
    [1, 1],
    [1, -1]
])

b = np.array([10, 2])

np.linalg.solve(A, b)
````

## Dengan Library: SymPy

SymPy bisa menyelesaikan sistem persamaan dengan bentuk simbolik.

````python
import sympy as sp

x, y = sp.symbols('x y')

persamaan_1 = sp.Eq(x + y, 10)
persamaan_2 = sp.Eq(x - y, 2)

sp.solve((persamaan_1, persamaan_2), (x, y))
````

## Catatan

Untuk sistem kecil, eliminasi manual membantu memahami idenya. Untuk sistem yang lebih besar, bentuk matriks dan library seperti NumPy jauh lebih praktis.
