# Variabel dan Ekspresi

Dalam matematika, variabel adalah simbol yang mewakili nilai. Biasanya kita melihat huruf seperti `x`, `y`, atau `n`.

Dalam Python, variabel juga dipakai untuk menyimpan nilai. Bedanya, variabel Python langsung punya isi tertentu saat program berjalan.

## Variabel di Matematika dan Python

Dalam matematika:

```text
x + 3
```

`x` bisa dianggap sebagai nilai yang belum diketahui.

Dalam Python:

```python
x = 5
x + 3
```

`x` sudah diberi nilai, jadi Python bisa langsung menghitung hasilnya.

````python
x = 5
hasil = x + 3
hasil
````

## Ekspresi

Ekspresi adalah gabungan angka, variabel, dan operasi.

Contoh:

```text
2x + 3
```

Di Python, perkalian harus ditulis eksplisit dengan `*`. Jadi `2x + 3` ditulis menjadi `2 * x + 3`.

````python
x = 10
2 * x + 3
````

## Substitusi Nilai

Substitusi berarti mengganti variabel dengan nilai tertentu.

Contoh:

```text
2x + 3
jika x = 10
2(10) + 3 = 23
```

````python
def hitung_ekspresi(x):
    return 2 * x + 3


hitung_ekspresi(10)
````

## Dengan Library: SymPy

Kalau ingin bekerja dengan simbol seperti di matematika, kita bisa memakai SymPy.

SymPy memungkinkan Python memahami `x` sebagai simbol, bukan sebagai variabel yang harus langsung punya angka.

````python
import sympy as sp

x = sp.Symbol('x')
ekspresi = 2 * x + 3
ekspresi
````

````python
ekspresi.subs(x, 10)
````

## Catatan

Untuk perhitungan biasa, variabel Python sudah cukup. Untuk aljabar simbolik seperti menyederhanakan ekspresi, menyelesaikan persamaan, atau mencari bentuk umum, SymPy lebih cocok.
