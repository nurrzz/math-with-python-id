# Persen

Persen berarti per seratus.

```text
50% = 50 per 100 = 50/100 = 0.5
25% = 25 per 100 = 25/100 = 0.25
```

Persen sering dipakai untuk diskon, kenaikan harga, nilai ujian, bunga, pajak, dan perbandingan data.

## Mengubah Persen ke Desimal

Untuk mengubah persen menjadi desimal, bagi dengan 100.

````python
persen = 25
desimal = persen / 100
desimal
````

## Menghitung Nilai dari Persen

Jika ingin mencari 20% dari 150000:

```text
20% × 150000 = 0.2 × 150000 = 30000
```

````python
harga = 150000
diskon_persen = 20

diskon = (diskon_persen / 100) * harga
diskon
````

## Contoh: Harga Setelah Diskon

Setelah diskon dihitung, harga akhir adalah harga awal dikurangi diskon.

````python
harga_awal = 150000
diskon_persen = 20

diskon = (diskon_persen / 100) * harga_awal
harga_akhir = harga_awal - diskon

harga_akhir
````

## Kenaikan Persen

Persen juga bisa dipakai untuk menghitung kenaikan.

Contoh: harga naik 10%.

````python
harga_awal = 100000
kenaikan_persen = 10

kenaikan = (kenaikan_persen / 100) * harga_awal
harga_baru = harga_awal + kenaikan

harga_baru
````

## Membuat Fungsi

Agar lebih rapi, perhitungan persen bisa dibungkus menjadi fungsi.

````python
def hitung_persen(persen, nilai):
    return (persen / 100) * nilai


def harga_setelah_diskon(harga_awal, diskon_persen):
    return harga_awal - hitung_persen(diskon_persen, harga_awal)


harga_setelah_diskon(150000, 20)
````

## Catatan

Persen terlihat sederhana, tapi penting untuk membaca perubahan. Dalam data, kenaikan dari 10 ke 20 adalah naik 100%, sedangkan kenaikan dari 1000 ke 1010 hanya naik 1%. Angka absolut dan persentase bisa memberi cerita yang berbeda.
