# Operasi Dasar

Operasi dasar adalah fondasi dari banyak konsep matematika lain. Sebelum masuk ke aljabar, statistika, atau kalkulus, kita perlu nyaman dulu dengan penjumlahan, pengurangan, perkalian, pembagian, dan urutan operasi.

Di Python, operasi dasar ditulis dengan simbol yang mirip dengan kalkulator, tetapi tetap mengikuti aturan matematika.

## Simbol Operasi di Python

| Operasi | Matematika | Python |
| --- | --- | --- |
| Penjumlahan | 5 + 2 | `5 + 2` |
| Pengurangan | 5 - 2 | `5 - 2` |
| Perkalian | 5 × 2 | `5 * 2` |
| Pembagian | 5 ÷ 2 | `5 / 2` |
| Pembagian bulat | 5 dibagi 2, ambil hasil bulat | `5 // 2` |
| Sisa bagi | sisa dari 5 ÷ 2 | `5 % 2` |

```python
print(5 + 2)
print(5 - 2)
print(5 * 2)
print(5 / 2)
print(5 // 2)
print(5 % 2)
```

## Pembagian Biasa, Pembagian Bulat, dan Sisa Bagi

Pembagian biasa memakai `/` dan hasilnya bisa desimal.

Pembagian bulat memakai `//` dan hanya mengambil bagian bilangan bulatnya.

Sisa bagi memakai `%` dan memberi tahu sisa setelah pembagian.

```python
jumlah_kue = 17
jumlah_orang = 5

print("Setiap orang dapat:", jumlah_kue // jumlah_orang)
print("Sisa kue:", jumlah_kue % jumlah_orang)
```

## Urutan Operasi

Python mengikuti urutan operasi matematika:

1. Tanda kurung
2. Pangkat
3. Perkalian dan pembagian
4. Penjumlahan dan pengurangan

Kalau ingin membuat maksud perhitungan lebih jelas, gunakan tanda kurung.

```python
hasil_1 = 2 + 3 * 4
hasil_2 = (2 + 3) * 4

print(hasil_1)
print(hasil_2)
```

## Membungkus Perhitungan ke Fungsi

Fungsi membuat perhitungan bisa dipakai ulang dan lebih mudah dibaca.

```python
def total_harga(harga_barang, jumlah_barang):
    return harga_barang * jumlah_barang


total_harga(15000, 3)
```

## Catatan

Operasi dasar terlihat sederhana, tapi konsep seperti `//` dan `%` sering dipakai dalam coding. Contohnya untuk membagi halaman, mengecek bilangan genap, membuat pola, atau mengatur indeks data.
