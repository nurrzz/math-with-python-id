# Pangkat dan Akar

Pangkat digunakan untuk mengalikan angka dengan dirinya sendiri beberapa kali.

```text
2^3 = 2 × 2 × 2 = 8
```

Akar adalah kebalikan dari pangkat.

```text
√9 = 3, karena 3 × 3 = 9
```

## Pangkat di Python

Python memakai operator `**` untuk pangkat.

````python
print(2 ** 3)
print(5 ** 2)
print(10 ** 0)
````

## Akar di Python

Akar kuadrat bisa dihitung dengan pangkat `0.5`.

````python
akar_9 = 9 ** 0.5
akar_16 = 16 ** 0.5

print(akar_9)
print(akar_16)
````

## Dengan Library Bawaan `math`

Python punya module bawaan `math` yang menyediakan fungsi `sqrt()` untuk akar kuadrat.

````python
import math

print(math.sqrt(9))
print(math.sqrt(16))
````

## Pangkat Pecahan

Akar juga bisa ditulis sebagai pangkat pecahan.

```text
√x = x^(1/2)
akar pangkat tiga dari x = x^(1/3)
```

````python
print(27 ** (1/3))
print(81 ** (1/4))
````

## Membuat Fungsi

Kita bisa membuat fungsi kecil untuk menghitung pangkat dan akar.

````python
def kuadrat(angka):
    return angka ** 2


def akar_kuadrat(angka):
    return angka ** 0.5


print(kuadrat(7))
print(akar_kuadrat(49))
````

## Catatan

Pangkat dan akar banyak muncul di statistika, geometri, aljabar, dan machine learning. Contohnya, standar deviasi memakai akar kuadrat dari varians.
