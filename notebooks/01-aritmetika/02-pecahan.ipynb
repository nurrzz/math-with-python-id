# Pecahan

Pecahan digunakan untuk menyatakan bagian dari keseluruhan.

Contoh sederhana:

```text
1/2 berarti 1 bagian dari 2 bagian yang sama besar.
3/4 berarti 3 bagian dari 4 bagian yang sama besar.
```

## Ide Matematika

Pecahan punya dua bagian:

```text
pembilang / penyebut
```

Pembilang adalah angka di atas. Penyebut adalah angka di bawah.

Contoh `3/4`:

- `3` adalah pembilang
- `4` adalah penyebut

````python
setengah = 1 / 2
tiga_perempat = 3 / 4

print(setengah)
print(tiga_perempat)
````

## Operasi Pecahan

Python akan langsung mengubah pembagian menjadi angka desimal jika kita memakai `/`. Untuk banyak kebutuhan praktis, ini cukup.

Tapi kalau ingin mempertahankan bentuk pecahan seperti `1/2`, kita bisa memakai module bawaan `fractions`.

````python
from fractions import Fraction

a = Fraction(1, 2)
b = Fraction(1, 4)

print(a + b)
print(a - b)
print(a * b)
print(a / b)
````

## Menyederhanakan Pecahan

Salah satu keuntungan memakai `Fraction` adalah Python otomatis menyederhanakan pecahan.

````python
Fraction(2, 4)
````

## Cara Manual: FPB

Untuk menyederhanakan pecahan secara manual, kita bisa membagi pembilang dan penyebut dengan FPB atau faktor persekutuan terbesar.

Contoh:

```text
2/4
FPB dari 2 dan 4 adalah 2
2 ÷ 2 = 1
4 ÷ 2 = 2
2/4 = 1/2
```

````python
def fpb(a, b):
    while b != 0:
        a, b = b, a % b
    return a


def sederhanakan_pecahan(pembilang, penyebut):
    pembagi = fpb(pembilang, penyebut)
    return pembilang // pembagi, penyebut // pembagi


sederhanakan_pecahan(2, 4)
````

## Catatan

Dalam data analysis, pecahan sering muncul sebagai rasio, proporsi, atau bagian dari total. Dalam Python, kita bisa memilih bentuk desimal untuk perhitungan praktis, atau `Fraction` saat ingin bentuk pecahan tetap terlihat.
