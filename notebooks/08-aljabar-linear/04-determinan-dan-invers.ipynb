# Determinan dan Invers Matriks

Determinan adalah nilai khusus dari matriks persegi. Salah satu kegunaannya adalah memberi tahu apakah sebuah matriks punya invers.

Invers matriks bisa dibayangkan seperti kebalikan dari matriks. Jika `A` dikalikan inversnya, hasilnya menjadi matriks identitas.

## Matriks Identitas

Matriks identitas mirip angka `1` dalam perkalian biasa.

```text
[1 0]
[0 1]
```

Jika sebuah matriks dikalikan dengan matriks identitas, nilainya tetap sama.

````python
import numpy as np

I = np.eye(2)
I
````

## Determinan Matriks 2 × 2

Untuk matriks:

```text
[a b]
[c d]
```

Determinan dihitung dengan:

```text
ad - bc
```

````python
def determinan_2x2(matriks):
    a = matriks[0][0]
    b = matriks[0][1]
    c = matriks[1][0]
    d = matriks[1][1]

    return a * d - b * c


A = [
    [1, 2],
    [3, 4]
]

determinan_2x2(A)
````

## Dengan Library: NumPy

NumPy menyediakan `np.linalg.det()` untuk menghitung determinan.

````python
A = np.array([
    [1, 2],
    [3, 4]
])

np.linalg.det(A)
````

## Invers Matriks

Jika determinan tidak sama dengan nol, matriks persegi biasanya punya invers.

````python
A_inv = np.linalg.inv(A)
A_inv
````

## Mengecek Invers

Jika `A` dikalikan dengan inversnya, hasilnya mendekati matriks identitas.

````python
A @ A_inv
````

## Hubungan dengan Sistem Persamaan

Invers matriks bisa dipakai untuk menyelesaikan sistem persamaan linear.

Jika:

```text
A x = b
```

Maka secara konsep:

```text
x = A⁻¹ b
```

````python
A = np.array([
    [1, 1],
    [1, -1]
])

b = np.array([10, 2])

x = np.linalg.inv(A) @ b
x
````

## Catatan

Dalam praktik komputasi, `np.linalg.solve()` sering lebih disarankan daripada menghitung invers secara eksplisit. Tapi memahami invers tetap penting untuk membaca ide matematikanya.
