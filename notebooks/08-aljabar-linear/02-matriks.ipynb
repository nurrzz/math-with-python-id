# Matriks

Matriks adalah susunan angka dalam baris dan kolom.

Contoh matriks 2 × 2:

```text
[1 2]
[3 4]
```

Matriks bisa dipakai untuk menyimpan data tabel, transformasi, sistem persamaan, dan banyak operasi numerik lain.

## Matriks di Python

Tanpa library, matriks bisa ditulis sebagai list di dalam list.

````python
matriks = [
    [1, 2],
    [3, 4]
]

matriks
````

## Mengakses Elemen Matriks

Karena matriks ditulis sebagai list di dalam list, kita bisa mengakses elemen dengan indeks baris dan kolom.

```python
matriks[baris][kolom]
```

Ingat, indeks Python dimulai dari `0`.

````python
matriks[0][0], matriks[0][1], matriks[1][0], matriks[1][1]
````

## Penjumlahan Matriks

Dua matriks bisa dijumlahkan jika ukurannya sama. Elemen yang posisinya sama dijumlahkan.

````python
A = [
    [1, 2],
    [3, 4]
]

B = [
    [10, 20],
    [30, 40]
]

hasil = []

for i in range(len(A)):
    baris = []
    for j in range(len(A[0])):
        baris.append(A[i][j] + B[i][j])
    hasil.append(baris)

hasil
````

## Dengan Library: NumPy

NumPy membuat matriks lebih mudah dibaca dan dihitung.

````python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [10, 20],
    [30, 40]
])

A + B
````

## Perkalian Elemen per Elemen

Dengan NumPy, operator `*` pada dua array melakukan perkalian elemen per elemen, bukan perkalian matriks matematika.

````python
A * B
````

## Catatan

Matriks sering menjadi bentuk dasar data tabular. Namun dalam matematika, matriks juga punya aturan operasi khusus, terutama untuk perkalian matriks.
