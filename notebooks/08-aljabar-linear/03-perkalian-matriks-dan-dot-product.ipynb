# Perkalian Matriks dan Dot Product

Perkalian matriks berbeda dari perkalian elemen per elemen. Ide dasarnya adalah menjumlahkan hasil kali antar komponen.

Salah satu operasi paling penting di sini adalah dot product.

## Dot Product

Untuk dua vektor:

```text
a = [2, 3]
b = [4, 5]
```

Dot product dihitung seperti ini:

```text
(2 × 4) + (3 × 5) = 8 + 15 = 23
```

````python
a = [2, 3]
b = [4, 5]

total = 0

for i in range(len(a)):
    total += a[i] * b[i]

total
````

## Membuat Fungsi Dot Product

Dot product hanya bisa dilakukan jika dua vektor punya panjang yang sama.

````python
def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError('Panjang vektor harus sama.')

    total = 0

    for i in range(len(a)):
        total += a[i] * b[i]

    return total


dot_product([2, 3], [4, 5])
````

## Dengan Library: NumPy

NumPy menyediakan `np.dot()` atau operator `@` untuk dot product dan perkalian matriks.

````python
import numpy as np

a = np.array([2, 3])
b = np.array([4, 5])

print(np.dot(a, b))
print(a @ b)
````

## Perkalian Matriks

Untuk matriks, operator `@` melakukan perkalian matriks sesuai aturan aljabar linear.

````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

A @ B
````

## Intuisi dalam Data

Dot product sering muncul saat menggabungkan nilai fitur dengan bobot.

Contoh sederhana:

```text
skor = fitur_1 × bobot_1 + fitur_2 × bobot_2 + fitur_3 × bobot_3
```

Ini adalah bentuk dot product.

````python
fitur = np.array([80, 70, 90])
bobot = np.array([0.4, 0.3, 0.3])

fitur @ bobot
````

## Catatan

Dot product adalah operasi kecil yang efeknya besar. Banyak model machine learning menggunakan ide ini sebagai bagian dari perhitungan prediksi.
