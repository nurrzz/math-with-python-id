# Vektor

Vektor adalah kumpulan angka yang punya urutan. Dalam Python, vektor bisa dibayangkan sebagai list angka.

Contoh:

```text
[2, 3]
```

Vektor ini bisa dibaca sebagai dua komponen: komponen pertama `2`, komponen kedua `3`.

## Vektor sebagai Data

Dalam data dan machine learning, satu objek sering direpresentasikan sebagai vektor.

Contoh data rumah:

```text
[luas, jumlah_kamar, harga]
[120, 3, 750000000]
```

Artinya, beberapa informasi disimpan dalam satu struktur angka.

````python
vektor = [2, 3]
vektor
````

## Penjumlahan Vektor

Dua vektor bisa dijumlahkan jika jumlah komponennya sama.

```text
[2, 3] + [4, 5] = [6, 8]
```

Komponen pertama dijumlahkan dengan komponen pertama, komponen kedua dengan komponen kedua.

````python
a = [2, 3]
b = [4, 5]

hasil = []

for i in range(len(a)):
    hasil.append(a[i] + b[i])

hasil
````

## Membuat Fungsi Penjumlahan Vektor

Agar bisa dipakai ulang, operasi tadi bisa dibuat sebagai fungsi.

````python
def tambah_vektor(a, b):
    if len(a) != len(b):
        raise ValueError('Panjang vektor harus sama.')

    hasil = []

    for i in range(len(a)):
        hasil.append(a[i] + b[i])

    return hasil


tambah_vektor([2, 3], [4, 5])
````

## Dengan Library: NumPy

NumPy membuat operasi vektor jauh lebih ringkas.

````python
import numpy as np

a = np.array([2, 3])
b = np.array([4, 5])

a + b
````

## Perkalian Skalar

Skalar adalah satu angka biasa. Jika vektor dikalikan skalar, setiap komponennya dikalikan angka tersebut.

```text
2 × [3, 4] = [6, 8]
```

````python
vektor = np.array([3, 4])
2 * vektor
````

## Catatan

Vektor adalah salah satu bentuk paling dasar dalam aljabar linear. Banyak data numerik bisa dibaca sebagai vektor, sehingga operasi vektor sangat penting dalam analisis data dan machine learning.
