# Probabilitas Bersyarat

Probabilitas bersyarat adalah peluang suatu kejadian terjadi setelah kita mengetahui informasi tertentu.

Biasanya ditulis seperti ini:

```text
P(A | B)
```

Dibaca: peluang A terjadi, jika B sudah diketahui terjadi.

## Contoh Intuisi

Misalnya ada data siswa dengan informasi apakah mereka belajar dan apakah mereka lulus.

Pertanyaan biasa:

```text
Berapa peluang siswa lulus?
```

Pertanyaan bersyarat:

```text
Berapa peluang siswa lulus jika diketahui siswa itu belajar?
```

````python
data = [
    {'belajar': True, 'lulus': True},
    {'belajar': True, 'lulus': True},
    {'belajar': True, 'lulus': False},
    {'belajar': False, 'lulus': True},
    {'belajar': False, 'lulus': False},
]

data
````

## Menghitung Secara Manual

Untuk menghitung peluang lulus jika belajar:

1. Ambil hanya data siswa yang belajar.
2. Dari kelompok itu, hitung berapa yang lulus.
3. Bagi jumlah yang lulus dengan jumlah siswa yang belajar.

````python
siswa_belajar = []

for siswa in data:
    if siswa['belajar'] == True:
        siswa_belajar.append(siswa)

lulus_dan_belajar = []

for siswa in siswa_belajar:
    if siswa['lulus'] == True:
        lulus_dan_belajar.append(siswa)

peluang_lulus_jika_belajar = len(lulus_dan_belajar) / len(siswa_belajar)
peluang_lulus_jika_belajar
````

## Membuat Fungsi

Kita bisa membuat fungsi sederhana untuk menghitung probabilitas bersyarat dari data dictionary.

````python
def probabilitas_bersyarat(data, kondisi, kejadian):
    data_sesuai_kondisi = []

    for item in data:
        if kondisi(item):
            data_sesuai_kondisi.append(item)

    data_sesuai_kejadian = []

    for item in data_sesuai_kondisi:
        if kejadian(item):
            data_sesuai_kejadian.append(item)

    return len(data_sesuai_kejadian) / len(data_sesuai_kondisi)


probabilitas_bersyarat(
    data,
    kondisi=lambda siswa: siswa['belajar'] == True,
    kejadian=lambda siswa: siswa['lulus'] == True
)
````

## Catatan tentang Data

Untuk dataset yang lebih besar, probabilitas bersyarat sering dihitung dengan tools data seperti pandas. Di notebook ini, kita sengaja memakai list dan dictionary dulu agar proses berpikirnya terlihat.

## Catatan

Probabilitas bersyarat penting karena informasi tambahan bisa mengubah peluang. Dalam dunia nyata, keputusan sering dibuat setelah kondisi tertentu diketahui.
