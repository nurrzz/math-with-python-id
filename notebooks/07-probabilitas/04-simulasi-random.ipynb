# Simulasi Random

Simulasi random digunakan untuk memperkirakan peluang dengan cara mencoba banyak percobaan secara acak.

Alih-alih hanya menghitung rumus, kita bisa membuat komputer melakukan percobaan berkali-kali, lalu melihat hasil rata-ratanya.

## Simulasi Lempar Koin

Peluang muncul kepala pada koin ideal adalah `1/2` atau `0.5`.

Kita bisa mensimulasikan lempar koin berkali-kali dan melihat apakah hasilnya mendekati `0.5`.

```python
import random

jumlah_percobaan = 1000
jumlah_kepala = 0

for _ in range(jumlah_percobaan):
    hasil = random.choice(['kepala', 'ekor'])

    if hasil == 'kepala':
        jumlah_kepala += 1

peluang_kepala = jumlah_kepala / jumlah_percobaan
peluang_kepala
```

## Semakin Banyak Percobaan

Semakin banyak percobaan, hasil simulasi biasanya semakin mendekati peluang teoritis.

```python
for jumlah_percobaan in [10, 100, 1000, 10000]:
    jumlah_kepala = 0

    for _ in range(jumlah_percobaan):
        hasil = random.choice(['kepala', 'ekor'])
        if hasil == 'kepala':
            jumlah_kepala += 1

    print(jumlah_percobaan, jumlah_kepala / jumlah_percobaan)
```

## Simulasi Lempar Dadu

Kita juga bisa mensimulasikan dadu untuk memperkirakan peluang muncul angka 6.

```python
jumlah_percobaan = 10000
jumlah_enam = 0

for _ in range(jumlah_percobaan):
    hasil = random.randint(1, 6)

    if hasil == 6:
        jumlah_enam += 1

jumlah_enam / jumlah_percobaan
```

## Dengan Library: NumPy

NumPy bisa membuat simulasi random lebih ringkas dan cepat untuk jumlah percobaan besar.

```python
import numpy as np

jumlah_percobaan = 10000
hasil_dadu = np.random.randint(1, 7, size=jumlah_percobaan)

np.mean(hasil_dadu == 6)
```

## Visualisasi Hasil Simulasi

Kita bisa melihat distribusi hasil dadu dengan grafik batang.

```python
import matplotlib.pyplot as plt

nilai, frekuensi = np.unique(hasil_dadu, return_counts=True)

plt.bar(nilai, frekuensi)
plt.title('Hasil Simulasi Lempar Dadu')
plt.xlabel('Angka dadu')
plt.ylabel('Frekuensi')
plt.grid(axis='y')
plt.show()
```

## Catatan

Simulasi random membantu kita memahami peluang lewat percobaan. Ini sangat berguna saat rumus terlalu rumit, atau saat kita ingin melihat perilaku sistem secara praktis.
