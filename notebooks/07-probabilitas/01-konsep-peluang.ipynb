# Konsep Peluang

Peluang digunakan untuk mengukur kemungkinan sebuah kejadian terjadi.

Nilai peluang berada di antara `0` dan `1`.

```text
0 berarti tidak mungkin terjadi
1 berarti pasti terjadi
```

Jika ditulis sebagai persen, peluang juga bisa dibaca dari `0%` sampai `100%`.

## Rumus Dasar

Jika semua hasil sama-sama mungkin terjadi, peluang bisa dihitung dengan:

```text
peluang = jumlah hasil yang diinginkan / jumlah semua kemungkinan hasil
```

Contoh dadu:

```text
P(muncul angka 6) = 1 / 6
```

````python
hasil_diinginkan = 1
semua_kemungkinan = 6

peluang = hasil_diinginkan / semua_kemungkinan
peluang
````

## Peluang Lebih dari Satu Hasil

Misalnya kita ingin menghitung peluang muncul angka genap pada dadu.

Angka genap pada dadu adalah `2`, `4`, dan `6`. Ada 3 hasil yang diinginkan dari total 6 kemungkinan.

```text
P(angka genap) = 3 / 6 = 0.5
```

````python
angka_dadu = [1, 2, 3, 4, 5, 6]
angka_genap = []

for angka in angka_dadu:
    if angka % 2 == 0:
        angka_genap.append(angka)

peluang_genap = len(angka_genap) / len(angka_dadu)
peluang_genap
````

## Membuat Fungsi Peluang

Kita bisa membuat fungsi sederhana untuk menghitung peluang dari daftar semua kemungkinan dan kondisi tertentu.

````python
def hitung_peluang(semua_hasil, kondisi):
    hasil_diinginkan = []

    for hasil in semua_hasil:
        if kondisi(hasil):
            hasil_diinginkan.append(hasil)

    return len(hasil_diinginkan) / len(semua_hasil)


hitung_peluang([1, 2, 3, 4, 5, 6], lambda angka: angka % 2 == 0)
````

## Dengan Library: NumPy

NumPy bisa membantu menghitung kondisi pada banyak data dengan lebih ringkas.

````python
import numpy as np

angka_dadu = np.array([1, 2, 3, 4, 5, 6])
peluang_genap = np.mean(angka_dadu % 2 == 0)
peluang_genap
````
