# Permutasi dan Kombinasi

Permutasi dan kombinasi digunakan untuk menghitung banyaknya cara memilih atau menyusun sesuatu.

Perbedaannya sederhana:

```text
Permutasi: urutan penting
Kombinasi: urutan tidak penting
```

## Faktorial

Sebelum masuk ke permutasi dan kombinasi, kita perlu mengenal faktorial.

```text
5! = 5 × 4 × 3 × 2 × 1 = 120
```

````python
def faktorial(n):
    hasil = 1

    for angka in range(1, n + 1):
        hasil *= angka

    return hasil


faktorial(5)
````

## Permutasi

Permutasi menghitung banyak susunan jika urutan penting.

Rumus permutasi memilih `r` dari `n` item:

```text
nPr = n! / (n - r)!
```

Contoh: memilih juara 1 dan 2 dari 5 orang. Urutan penting, karena juara 1 dan juara 2 berbeda.

````python
def permutasi(n, r):
    return faktorial(n) / faktorial(n - r)


permutasi(5, 2)
````

## Kombinasi

Kombinasi menghitung banyak pilihan jika urutan tidak penting.

Rumus kombinasi memilih `r` dari `n` item:

```text
nCr = n! / (r! × (n - r)!)
```

Contoh: memilih 2 orang dari 5 orang untuk menjadi anggota tim. Urutan tidak penting.

````python
def kombinasi(n, r):
    return faktorial(n) / (faktorial(r) * faktorial(n - r))


kombinasi(5, 2)
````

## Dengan Library: `math`

Python punya module bawaan `math` untuk faktorial, permutasi, dan kombinasi.

````python
import math

print(math.factorial(5))
print(math.perm(5, 2))
print(math.comb(5, 2))
````

## Catatan

Gunakan permutasi saat posisi atau urutan penting. Gunakan kombinasi saat yang penting hanya siapa atau apa yang terpilih, bukan urutannya.
