# Grafik Trigonometri

Fungsi trigonometri tidak hanya dipakai pada segitiga. `sin` dan `cos` juga membentuk pola gelombang.

Pola ini sering muncul dalam suara, getaran, sinyal, gerakan berulang, dan visualisasi periodik.

## Pola Berulang

Fungsi `sin(x)` dan `cos(x)` berulang setiap `2π` radian.

Artinya, setelah satu putaran penuh, bentuk grafiknya kembali mengulang.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, label='sin(x)')
plt.plot(x, y_cos, label='cos(x)')
plt.axhline(0, color='black', linewidth=0.8)
plt.title('Grafik sin(x) dan cos(x)')
plt.xlabel('x dalam radian')
plt.ylabel('nilai fungsi')
plt.grid(True)
plt.legend()
plt.show()
````

## Amplitudo

Amplitudo adalah tinggi gelombang dari garis tengah.

Pada fungsi:

```text
y = A sin(x)
```

Nilai `A` mengatur tinggi gelombang.

````python
x = np.linspace(0, 2 * np.pi, 200)

for A in [1, 2, 0.5]:
    y = A * np.sin(x)
    plt.plot(x, y, label=f'A = {A}')

plt.axhline(0, color='black', linewidth=0.8)
plt.title('Pengaruh Amplitudo pada sin(x)')
plt.xlabel('x dalam radian')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
````

## Periode

Periode adalah panjang satu pengulangan gelombang. Untuk `sin(x)`, periode dasarnya adalah `2π`.

Jika fungsi ditulis:

```text
y = sin(Bx)
```

Nilai `B` memengaruhi seberapa cepat gelombang berulang.

````python
x = np.linspace(0, 2 * np.pi, 200)

for B in [1, 2, 3]:
    y = np.sin(B * x)
    plt.plot(x, y, label=f'B = {B}')

plt.axhline(0, color='black', linewidth=0.8)
plt.title('Pengaruh B pada Periode sin(Bx)')
plt.xlabel('x dalam radian')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
````

## Catatan

Grafik trigonometri membantu kita melihat bahwa `sin` dan `cos` bukan sekadar rumus segitiga. Keduanya juga menggambarkan pola berulang, dan pola seperti ini sangat sering muncul di teknologi, data, audio, sensor, dan machine learning.
