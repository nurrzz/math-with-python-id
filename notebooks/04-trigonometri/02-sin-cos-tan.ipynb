# Sin, Cos, dan Tan

`sin`, `cos`, dan `tan` adalah tiga fungsi dasar dalam trigonometri.

Dalam segitiga siku-siku, ketiganya menghubungkan sudut dengan perbandingan panjang sisi.

## Ide Matematika

Untuk sebuah sudut pada segitiga siku-siku:

```text
sin(sudut) = sisi depan / sisi miring
cos(sudut) = sisi samping / sisi miring
tan(sudut) = sisi depan / sisi samping
```

Perbandingan ini disebut rasio, karena membandingkan dua panjang sisi.

## Menghitung dengan Python

Module `math` menyediakan `sin()`, `cos()`, dan `tan()`. Input sudutnya harus dalam radian.

````python
import math

sudut_derajat = 45
sudut_radian = math.radians(sudut_derajat)

print(math.sin(sudut_radian))
print(math.cos(sudut_radian))
print(math.tan(sudut_radian))
````

## Beberapa Nilai Penting

Beberapa sudut sering muncul dalam trigonometri dasar: `0°`, `30°`, `45°`, `60°`, dan `90°`.

Kita bisa membuat tabel kecil dengan Python.

````python
sudut_derajat = [0, 30, 45, 60, 90]

for derajat in sudut_derajat:
    radian = math.radians(derajat)
    print(derajat, round(math.sin(radian), 4), round(math.cos(radian), 4))
````

## Dengan Library: NumPy

NumPy berguna saat kita ingin menghitung banyak nilai sekaligus.

````python
import numpy as np

sudut_derajat = np.array([0, 30, 45, 60, 90])
sudut_radian = np.radians(sudut_derajat)

print(np.sin(sudut_radian))
print(np.cos(sudut_radian))
````

## Catatan

Nilai seperti `cos(90°)` kadang muncul sebagai angka sangat kecil, bukan benar-benar nol, misalnya `6.123e-17`. Itu terjadi karena komputer menghitung dengan pendekatan desimal. Secara konsep, nilainya tetap dianggap `0`.
