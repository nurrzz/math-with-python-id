# Identitas Trigonometri

Identitas trigonometri adalah persamaan yang selalu benar untuk sudut tertentu selama nilainya terdefinisi.

Salah satu identitas paling dasar adalah:

```text
sin²(x) + cos²(x) = 1
```

## Mengecek dengan Angka

Kita bisa mengecek identitas ini dengan beberapa nilai sudut. Karena Python memakai radian, sudut derajat perlu dikonversi dulu.

````python
import math

for derajat in [0, 30, 45, 60, 90]:
    x = math.radians(derajat)
    hasil = math.sin(x)**2 + math.cos(x)**2
    print(derajat, hasil)
````

## Efek Pembulatan

Hasil komputer bisa terlihat seperti `0.9999999999999999` atau `1.0000000000000002`. Ini bukan berarti identitasnya salah. Itu hanya efek perhitungan floating point.

````python
for derajat in [0, 30, 45, 60, 90]:
    x = math.radians(derajat)
    hasil = math.sin(x)**2 + math.cos(x)**2
    print(derajat, round(hasil, 10))
````

## Hubungan Tan, Sin, dan Cos

Identitas lain yang penting:

```text
tan(x) = sin(x) / cos(x)
```

Selama `cos(x)` tidak sama dengan nol, nilai `tan(x)` bisa dihitung dari `sin(x)` dan `cos(x)`.

````python
derajat = 45
x = math.radians(derajat)

tan_langsung = math.tan(x)
tan_dari_rasio = math.sin(x) / math.cos(x)

print(tan_langsung)
print(tan_dari_rasio)
````

## Dengan Library: SymPy

SymPy bisa menyederhanakan bentuk trigonometri secara simbolik.

````python
import sympy as sp

x = sp.Symbol('x')
ekspresi = sp.sin(x)**2 + sp.cos(x)**2

sp.trigsimp(ekspresi)
````

## Catatan

Identitas trigonometri membantu menyederhanakan rumus. Dalam coding, identitas ini juga membantu kita memahami kenapa beberapa rumus bisa ditulis dalam bentuk yang berbeda tetapi menghasilkan nilai yang sama.
