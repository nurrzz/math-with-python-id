# Sudut, Derajat, dan Radian

Trigonometri banyak berbicara tentang sudut. Sebelum memakai `sin`, `cos`, dan `tan`, kita perlu memahami dua satuan sudut yang sering dipakai: derajat dan radian.

Di sekolah, kita sering memakai derajat. Di Python dan banyak library matematika, sudut biasanya dihitung dalam radian.

## Derajat dan Radian

Satu putaran penuh sama dengan:

```text
360 derajat = 2π radian
```

Setengah putaran:

```text
180 derajat = π radian
```

Dari sini, kita bisa mengubah derajat ke radian dengan rumus:

```text
radian = derajat × π / 180
```

````python
pi = 3.141592653589793

derajat = 180
radian = derajat * pi / 180

radian
````

## Membuat Fungsi Konversi

Agar lebih mudah dipakai ulang, konversi derajat dan radian bisa dibuat sebagai fungsi.

````python
def derajat_ke_radian(derajat):
    pi = 3.141592653589793
    return derajat * pi / 180


def radian_ke_derajat(radian):
    pi = 3.141592653589793
    return radian * 180 / pi


print(derajat_ke_radian(180))
print(radian_ke_derajat(3.141592653589793))
````

## Dengan Library: `math`

Python punya module bawaan `math` untuk konversi sudut.

````python
import math

print(math.radians(180))
print(math.degrees(math.pi))
````

## Catatan

Saat memakai fungsi trigonometri di Python, biasakan mengecek apakah input sudutnya harus dalam derajat atau radian. Module `math` dan NumPy memakai radian.
