# Turunan Simbolik

Turunan simbolik berarti mencari bentuk rumus turunannya, bukan hanya menghitung nilai pendekatan di satu titik.

Contoh:

```text
f(x) = x²
f'(x) = 2x
```

## Aturan Pangkat

Salah satu aturan turunan paling dasar:

```text
jika f(x) = xⁿ, maka f'(x) = n xⁿ⁻¹
```

Contoh:

```text
f(x) = x²
f'(x) = 2x
```

````python
def turunan_x_kuadrat(x):
    return 2 * x


turunan_x_kuadrat(3)
````

## Dengan Library: SymPy

SymPy bisa mencari turunan simbolik dengan `diff()`.

````python
import sympy as sp

x = sp.Symbol('x')
f = x**2

turunan = sp.diff(f, x)
turunan
````

````python
turunan.subs(x, 3)
````

## Fungsi yang Lebih Panjang

SymPy juga bisa menurunkan fungsi yang lebih panjang.

````python
f = 3*x**3 - 2*x**2 + 5*x - 7
sp.diff(f, x)
````

## Turunan Fungsi Trigonometri

Beberapa fungsi trigonometri juga punya turunan yang sering muncul:

```text
d/dx sin(x) = cos(x)
d/dx cos(x) = -sin(x)
```

````python
sp.diff(sp.sin(x), x), sp.diff(sp.cos(x), x)
````

## Catatan

Turunan simbolik membantu saat kita ingin memahami bentuk umum perubahan sebuah fungsi. Turunan numerik membantu saat kita hanya butuh pendekatan nilai. Keduanya berguna, tergantung konteks.
