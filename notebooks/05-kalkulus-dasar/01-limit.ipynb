# Limit

Limit membantu kita memahami nilai yang didekati oleh sebuah fungsi saat input mendekati nilai tertentu.

Kata kuncinya adalah *mendekati*, bukan selalu sama dengan. Ini penting karena kadang fungsi sulit atau tidak bisa dihitung tepat di satu titik, tetapi perilakunya di sekitar titik itu masih bisa diamati.

## Intuisi Sederhana

Misalnya fungsi:

```text
f(x) = x + 1
```

Saat `x` mendekati `2`, nilai `f(x)` mendekati `3`.

```text
x = 1.9  -> f(x) = 2.9
x = 1.99 -> f(x) = 2.99
x = 2.01 -> f(x) = 3.01
x = 2.1  -> f(x) = 3.1
```

````python
def f(x):
    return x + 1


for x in [1.9, 1.99, 1.999, 2.001, 2.01, 2.1]:
    print(x, f(x))
````

## Contoh yang Lebih Menarik

Perhatikan fungsi:

```text
f(x) = (x² - 1) / (x - 1)
```

Jika `x = 1`, pembaginya menjadi `0`, sehingga fungsi tidak bisa dihitung langsung. Tapi kita tetap bisa melihat nilai fungsi saat `x` mendekati `1`.

````python
def g(x):
    return (x**2 - 1) / (x - 1)


for x in [0.9, 0.99, 0.999, 1.001, 1.01, 1.1]:
    print(x, g(x))
````

## Membaca Hasil

Nilai fungsi mendekati `2` saat `x` mendekati `1`. Jadi secara intuitif:

```text
limit dari (x² - 1) / (x - 1), saat x mendekati 1, adalah 2
```

Walaupun fungsi aslinya tidak bisa dihitung tepat di `x = 1`, limitnya tetap bisa ada.

## Dengan Library: SymPy

SymPy bisa menghitung limit secara simbolik.

````python
import sympy as sp

x = sp.Symbol('x')
ekspresi = (x**2 - 1) / (x - 1)

sp.limit(ekspresi, x, 1)
````

## Catatan

Limit adalah dasar penting untuk turunan dan integral. Di tahap awal, cukup pahami limit sebagai cara membaca perilaku fungsi saat input bergerak mendekati suatu nilai.
