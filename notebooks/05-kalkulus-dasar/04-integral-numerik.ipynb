# Integral Numerik

Integral bisa dipahami sebagai luas area di bawah kurva.

Jika turunan berbicara tentang perubahan, integral sering dipakai untuk menjumlahkan perubahan kecil menjadi total.

## Ide Luas dengan Persegi Panjang

Kita bisa memperkirakan luas area di bawah kurva dengan membaginya menjadi banyak persegi panjang kecil.

Semakin banyak persegi panjangnya, biasanya hasil pendekatannya semakin baik.

````python
def f(x):
    return x ** 2


def integral_numerik(f, a, b, n=1000):
    lebar = (b - a) / n
    total = 0

    for i in range(n):
        x = a + i * lebar
        total += f(x) * lebar

    return total


integral_numerik(f, 0, 1)
````

## Membaca Hasil

Untuk fungsi:

```text
f(x) = x²
```

Integral dari `0` sampai `1` bernilai `1/3`, atau sekitar `0.3333`.

Hasil numerik mendekati nilai tersebut.

````python
for n in [10, 100, 1000, 10000]:
    print(n, integral_numerik(f, 0, 1, n))
````

## Dengan Library: NumPy

NumPy menyediakan fungsi seperti `trapz` melalui `trapezoid()` untuk pendekatan integral dengan metode trapesium.

````python
import numpy as np

x_values = np.linspace(0, 1, 1000)
y_values = x_values ** 2

np.trapezoid(y_values, x_values)
````

## Visualisasi Area

Kita bisa menggambar area di bawah kurva untuk membuat ide integral lebih terasa.

````python
import matplotlib.pyplot as plt

x_values = np.linspace(0, 1, 200)
y_values = x_values ** 2

plt.plot(x_values, y_values, label='f(x) = x²')
plt.fill_between(x_values, y_values, alpha=0.3, label='area di bawah kurva')
plt.title('Integral sebagai Luas Area')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
````

## Dengan Library: SymPy

SymPy bisa menghitung integral secara simbolik.

````python
import sympy as sp

x = sp.Symbol('x')
sp.integrate(x**2, (x, 0, 1))
````

## Catatan

Integral numerik sering dipakai saat bentuk fungsi sulit diintegralkan secara manual, atau saat data yang kita punya berupa titik-titik hasil pengukuran, bukan rumus matematika yang rapi.
