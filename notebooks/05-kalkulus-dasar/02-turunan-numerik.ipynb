# Turunan Numerik

Turunan mengukur seberapa cepat sebuah fungsi berubah.

Jika fungsi digambar sebagai grafik, turunan di satu titik bisa dibayangkan sebagai kemiringan garis singgung di titik tersebut.

## Ide Kemiringan

Untuk dua titik pada grafik, kemiringan garis bisa dihitung dengan:

```text
perubahan y / perubahan x
```

Atau:

```text
(f(x + h) - f(x)) / h
```

Jika `h` dibuat sangat kecil, nilai ini mendekati turunan di titik `x`.

````python
def f(x):
    return x ** 2


def turunan_numerik(f, x, h=0.0001):
    return (f(x + h) - f(x)) / h


turunan_numerik(f, 3)
````

## Membaca Hasil

Untuk fungsi:

```text
f(x) = x²
```

Turunan matematisnya adalah:

```text
f'(x) = 2x
```

Jika `x = 3`, maka turunannya `6`. Hasil numerik mendekati angka tersebut.

## Pengaruh Nilai `h`

Nilai `h` yang lebih kecil biasanya membuat hasil lebih dekat, tetapi kalau terlalu kecil, komputer bisa terkena masalah pembulatan desimal.

````python
for h in [1, 0.1, 0.01, 0.0001, 0.000001]:
    print(h, turunan_numerik(f, 3, h))
````

## Visualisasi Fungsi

Grafik membantu kita melihat bahwa turunan berhubungan dengan kemiringan kurva.

````python
import numpy as np
import matplotlib.pyplot as plt

x_values = np.linspace(-5, 5, 200)
y_values = f(x_values)

plt.plot(x_values, y_values, label='f(x) = x²')
plt.scatter([3], [f(3)], color='red', label='titik x = 3')
plt.title('Fungsi f(x) = x²')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
````
