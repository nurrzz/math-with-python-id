# Fungsi Linear

Fungsi linear adalah fungsi yang grafiknya berbentuk garis lurus. Bentuk umumnya:

```text
f(x) = mx + b
```

`m` adalah kemiringan garis. `b` adalah titik potong dengan sumbu y.

## Contoh Fungsi Linear

Misalnya:

```text
f(x) = 2x + 1
```

Jika `x` bertambah 1, nilai `f(x)` bertambah 2. Angka `2` inilah yang disebut kemiringan.

````python
def f(x):
    return 2 * x + 1


for x in [-2, -1, 0, 1, 2]:
    print(x, f(x))
````

## Membuat Data untuk Grafik

Sebelum menggambar grafik, kita buat daftar nilai `x`, lalu hitung nilai `y` dari fungsi.

````python
x_values = list(range(-5, 6))
y_values = [f(x) for x in x_values]

print(x_values)
print(y_values)
````

## Dengan Library: Matplotlib

Matplotlib digunakan untuk membuat grafik di Python.

````python
import matplotlib.pyplot as plt

plt.plot(x_values, y_values, marker='o')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Grafik f(x) = 2x + 1')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
````

## Mengubah Kemiringan

Jika nilai `m` berubah, kemiringan garis juga berubah.

````python
def garis(x, m, b):
    return m * x + b

x_values = list(range(-5, 6))

for m in [1, 2, -1]:
    y_values = [garis(x, m, 0) for x in x_values]
    plt.plot(x_values, y_values, marker='o', label=f'm = {m}')

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Perbandingan Kemiringan Garis')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
````

## Catatan

Fungsi linear sering muncul dalam hubungan yang berubah secara tetap. Contohnya biaya parkir per jam, gaji per hari kerja, atau jarak dengan kecepatan konstan.
