# Konsep Fungsi

Fungsi adalah aturan yang menerima input, lalu menghasilkan output.

Dalam matematika, fungsi sering ditulis seperti ini:

```text
f(x) = 2x + 3
```

Artinya, setiap nilai `x` yang masuk akan dikalikan 2, lalu ditambah 3.

## Input dan Output

Kalau `f(x) = 2x + 3`, maka:

```text
f(1) = 2(1) + 3 = 5
f(2) = 2(2) + 3 = 7
f(3) = 2(3) + 3 = 9
```

Fungsi mengubah input menjadi output berdasarkan aturan tertentu.

````python
def f(x):
    return 2 * x + 3


print(f(1))
print(f(2))
print(f(3))
````

## Tabel Nilai

Salah satu cara memahami fungsi adalah membuat tabel input dan output.

````python
input_x = [-2, -1, 0, 1, 2]

for x in input_x:
    print(x, f(x))
````

## Membuat Banyak Output Sekaligus

Di Python, kita bisa menghitung output untuk banyak nilai `x` menggunakan list comprehension.

````python
x_values = [-2, -1, 0, 1, 2]
y_values = [f(x) for x in x_values]

y_values
````

## Dengan Library: SymPy

SymPy bisa menulis fungsi sebagai ekspresi simbolik. Ini berguna kalau kita ingin melihat bentuk matematikanya, bukan hanya hasil angka.

````python
import sympy as sp

x = sp.Symbol('x')
ekspresi = 2 * x + 3

ekspresi
````

````python
ekspresi.subs(x, 5)
````
