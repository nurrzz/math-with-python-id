# Fungsi Kuadrat

Fungsi kuadrat adalah fungsi yang memiliki variabel berpangkat dua. Bentuk umumnya:

```text
f(x) = ax² + bx + c
```

Grafiknya berbentuk parabola.

## Contoh Fungsi Kuadrat

Misalnya:

```text
f(x) = x²
```

Nilai output selalu positif atau nol karena hasil kuadrat tidak negatif.

````python
def f(x):
    return x ** 2


for x in [-3, -2, -1, 0, 1, 2, 3]:
    print(x, f(x))
````

## Menggambar Grafik Fungsi Kuadrat

Kita bisa membuat banyak titik `x`, menghitung `y`, lalu menggambarnya dengan Matplotlib.

````python
import matplotlib.pyplot as plt

x_values = list(range(-5, 6))
y_values = [f(x) for x in x_values]

plt.plot(x_values, y_values, marker='o')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Grafik f(x) = x²')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
````

## Pengaruh Nilai `a`

Pada fungsi:

```text
f(x) = ax²
```

Nilai `a` memengaruhi arah dan lebar parabola.

- Jika `a` positif, parabola terbuka ke atas.
- Jika `a` negatif, parabola terbuka ke bawah.
- Jika nilai mutlak `a` besar, parabola lebih sempit.

````python
x_values = list(range(-5, 6))

for a in [1, 2, -1]:
    y_values = [a * x**2 for x in x_values]
    plt.plot(x_values, y_values, marker='o', label=f'a = {a}')

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Pengaruh Nilai a pada Fungsi Kuadrat')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
````

## Dengan Library: SymPy

SymPy bisa membantu melihat bentuk akar dari fungsi kuadrat saat `f(x) = 0`.

````python
import sympy as sp

x = sp.Symbol('x')
ekspresi = x**2 - 5*x + 6

sp.solve(sp.Eq(ekspresi, 0), x)
````
