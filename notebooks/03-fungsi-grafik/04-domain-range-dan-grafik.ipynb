# Domain, Range, dan Grafik

Domain adalah kumpulan nilai input yang boleh masuk ke fungsi. Range adalah kumpulan nilai output yang mungkin dihasilkan.

Secara sederhana:

```text
domain = nilai x yang boleh dipakai
range = nilai y yang keluar dari fungsi
```

## Contoh Domain

Untuk fungsi:

```text
f(x) = x + 1
```

Hampir semua angka bisa menjadi input.

Tapi untuk fungsi:

```text
g(x) = 1 / x
```

`x = 0` tidak boleh dipakai, karena pembagian dengan nol tidak terdefinisi.

````python
def g(x):
    return 1 / x


for x in [-2, -1, 1, 2]:
    print(x, g(x))
````

## Menghindari Input yang Tidak Valid

Dalam kode, kita bisa memeriksa input sebelum menghitung.

````python
def g_aman(x):
    if x == 0:
        raise ValueError('x tidak boleh 0.')

    return 1 / x


g_aman(2)
````

## Range pada Fungsi Kuadrat

Untuk fungsi:

```text
f(x) = x²
```

Outputnya tidak pernah negatif. Jadi range-nya adalah nilai `y >= 0`.

````python
def f(x):
    return x ** 2

x_values = list(range(-5, 6))
y_values = [f(x) for x in x_values]

print(min(y_values))
print(max(y_values))
````

## Melihat Domain dan Range lewat Grafik

Grafik membantu kita melihat hubungan antara input dan output secara visual.

````python
import matplotlib.pyplot as plt

plt.plot(x_values, y_values, marker='o')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Domain dan Range pada f(x) = x²')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
````

## Catatan

Domain dan range membantu kita tidak hanya menghitung fungsi, tetapi juga memahami batas masuk dan keluar dari sebuah aturan. Ini penting saat membuat model, membaca grafik, atau memvalidasi input dalam program.
